```python
import pandas as pd
import numpy as np

# Create sample expense data
data = {
    "Date": pd.date_range(start="2024-01-01", periods=30),
    "Category": np.random.choice(
        ["Food", "Travel", "Shopping", "Entertainment", "Bills"], 30
    ),
    "Amount": np.random.randint(100, 2000, 30),
    "Payment_Mode": np.random.choice(
        ["Cash", "UPI", "Card"], 30
    )
}

df = pd.DataFrame(data)
df.head()
```

```python
df.tail()
```

```python
df.info()
```

```python
df['Amount'].sum()
```

```python
category_spending=df.groupby('Category')['Amount'].sum()
category_spending
```

```python
df.groupby('Category')['Amount'].mean()
```

```python
df.groupby('Payment_Mode')['Amount'].sum()
```

```python
category_spending.plot(kind='bar',title='Category-wise Spending')
```

```python
import matplotlib.pyplot as plt
plt.show()
```

## Expense Analysis Insights

- Highest spending category can be easily identified
- Food and Shopping contribute significantly to total expenses
- Digital payments (UPI/Card) dominate over cash
- Category-wise analysis helps in budgeting and control
